## 1. Variables & Types

Python is **dynamically typed** — you don't declare types, Python figures it out.
You can also reassign a variable to a completely different type (though this is usually bad practice).

````python
# Integers and floats
age = 17

# String
name = "Derick"

# Boolean
is_student = True

print(age,  name, is_student)
````

````python
# Check the type of a variable with type()
print(type(age))
print(type(name))
print(type(is_student))
````

````python
# You can dynamically change the type of a variable by assigning a new value of a different type (NOT RECOMMENDED!)
is_student = 3
````

````python
print(type(age))
print(type(name))
print(type(is_student)) # Notice now that is_student changed from a boolean to an integer type
````

## ⚠️ Why Changing a Variable's Type is a Bad Idea

In the example above, `is_student` started out as `True` (a **boolean** — a yes/no value), but then we reassigned it to `3` (an **integer** — a number). Python allowed it, but that doesn't mean we *should* do it!

Here's why professional developers avoid this:

- **It causes confusion.** If someone reads your code and sees a variable called `is_student`, they'll naturally assume it holds a `True` or `False` value — not a number. Changing its type mid-way through a program is surprising and hard to follow.

- **It leads to bugs.** Imagine code further down that does `if is_student:` — it would still "work" because `3` is truthy in Python, but the logic is no longer clear. Sneaky bugs like this can take hours to track down.

- **It makes teamwork harder.** In a real job, other developers (or future-you!) will read your code. Consistent types make it much easier to understand what a variable is supposed to hold.

**The golden rule:** Once you decide what a variable represents, keep it that way. If you need a different kind of value, make a new variable with a descriptive name instead.

> 💡 Many modern languages (like TypeScript, Java, and C#) actually *prevent* you from changing a variable's type — they'll throw an error. Python is flexible, but that flexibility is a responsibility, not a free pass!

````python
# Basic arithmetic — mostly what you'd expect
print(10 + 3)   # addition
print(10 - 3)   # subtraction
print(10 * 3)   # multiplication
print(10 / 3)   # division — always returns a float!
print(10 // 3)  # integer division (floor)
print(10 % 3)   # modulus (remainder)
print(10 ** 3)  # power (10 cubed)
````

## 2. Strings

Strings are one of the most-used types in Python. They come with lots of built-in methods.

````python
sentence = "  Hello, World!  "

print(sentence.upper())       # uppercase
print(sentence.lower())       # lowercase
print(sentence.strip())       # remove whitespace from edges
print(sentence.replace("World", "Python"))  # replace a word
print(len(sentence))          # length of string
````

````python
# f-strings: the best way to format strings in Python
name = "Alex"
score = 95.6786878434368768545867

print(f"Student: {name}")
print(f"Score: {score}")
print(f"{name} scored {score} on the test")
````

### Controlling decimal places with format specs

When working with numbers in f-strings, you'll often want to control *how* a number is displayed — for example, showing a price as `3.99` rather than `3.9900000000000002`, or rounding a test score to a whole number.

Python gives us a handy mini-language for this called a **format spec**. You add it inside the curly braces, after the variable name, separated by a colon:

```
f"{variable:.Xf}"
```

Breaking it down piece by piece:

| Part | Meaning |
|------|---------|
| `:`  | "I'm about to describe how to format this value" |
| `.X` | precision — how many digits to show after the decimal point |
| `f`  | the *type*: `f` stands for **fixed-point** (i.e. a regular decimal number) |

So `{score:.2f}` means *"format `score` as a fixed-point number with 2 decimal places"*. Let's see it in action:

````python
print(f"{score:.2f}")   # → 95.68  (2 decimal places)
print(f"{score:.1f}")   # → 95.7   (1 decimal place)
print(f"{score:.0f}")   # → 96     (rounded to whole number)
````

````python
print(f"{name} scored {score:.0f} out of 100")
````

### Format spec type letters

The letter at the end of a format spec controls *how* the value is interpreted. Here are the most useful ones:

| Letter | Name | Use case | Example |
|--------|------|----------|---------|
| `f` | Fixed-point float | Prices, scores, measurements | `{3.14159:.2f}` → `3.14` |
| `d` | Integer (decimal) | Counts, IDs — removes decimal point | `{42.0:d}` → error; use `{42:d}` |
| `e` | Scientific notation | Very large or very small numbers | `{0.000123:.2e}` → `1.23e-04` |
| `%` | Percentage | Rates, proportions (multiplies by 100) | `{0.856:.1%}` → `85.6%` |
| `,` | Thousands separator | Large numbers, currency | `{1000000:,}` → `1,000,000` |

You can **combine** a separator with a type and precision:
```
f"{value:,.2f}"   →  1,000,000.00
```

> 💡 The `,` is not a type letter — it's a *grouping option* that sits between the colon and the precision. You can stack it with `f`, `d`, or `e`.

````python
price = 1000000
ratio = 0.856

print(f"{price:,}")
print(f"{ratio:.1%}")
````

````python
# You can combine them too
print(f"R{price:,.2f}")
````

````python
# Scientific notation — useful when numbers are very large or very small
avogadro = 602214076000000000000000   # 6.022 × 10²³
electron_mass = 0.00000000000000000000000000000091093837015  # kg

print(f"Avogadro's number: {avogadro:.3e}")     # → 6.022e+23
print(f"Electron mass:     {electron_mass:.3e}") # → 9.109e-31

# Integer format — strips the decimal entirely
student_count = 42
print(f"Students in class: {student_count:d}")   # → 42

# Combining thousands separator with integers
population = 62000000
print(f"Population: {population:,d}")            # → 62,000,000
````

## 3. Truthy and Falsy

In Python, every value can be used in an `if` condition — not just `True` and `False`. Values that act like `True` are called **truthy**, and values that act like `False` are called **falsy**.

This matters a lot in real code — you'll use it constantly without even thinking about it!

Here are all the **falsy** values in Python (everything else is **truthy**):

| Value | Type |
|---|---|
| `False` | Boolean |
| `0` or `0.0` | Number (zero) |
| `""` | Empty string |
| `[]` | Empty list |
| `{}` | Empty dict |
| `None` | The "no value" value |


````python
# bool() shows you how Python evaluates any value
print(bool(False))   # False — obviously!
print(bool(0))       # False — zero is falsy
print(bool(42))      # True  — any non-zero number is truthy
print(bool(""))      # False — empty string is falsy
print(bool("hi"))    # True  — any non-empty string is truthy
print(bool([]))      # False — empty list is falsy
print(bool([1, 2]))  # True  — a list with items is truthy
print(bool(None))    # False — None is always falsy
````

````python
# You almost never write bool() yourself —
# Python checks truthiness automatically in if statements

username = "Alex"  # try changing to "Alex"

if username:
    print(f"Welcome back, {username}!")
else:
    print("No username provided.")

# Same for lists — check if there's anything in them
items_in_cart = ["apple", "milk"]  # try changing to ["apple", "milk"]

if items_in_cart:
    print(f"You have {len(items_in_cart)} item(s) in your cart.")
else:
    print("Your cart is empty.")
````

## 4. Lists

Lists are ordered, changeable collections. Think of them like dynamic arrays.
They can hold mixed types (though in data work, they usually hold one type).


````python
scores = [85, 92, 78, 95, 61, 88]

print(scores[0])
print(scores[-1])
print(scores[1:4])
print(scores[-3])
print(len(scores))
````

````python
# Splitting a string into a list
csv_line = "Alice,Math,87,pass"
parts = csv_line.split(",")
print(parts)
print(parts[0])  # first element
print(parts[-1]) # last element
````

````python
# Modifying lists
print(f"Original: {scores} \n")

scores.append(73)       # add to end
print(f"After append: {scores} \n")

scores.insert(0, 100)   # insert at position 0
print(f"After insert: {scores} \n")
scores.remove(61)       # remove first occurrence of 61
print(f"After remove: {scores} \n")

# Useful list functions
print(f"Min: {min(scores)}, Max: {max(scores)}, Sum: {sum(scores)}")
print(f"Average: {sum(scores) / len(scores):.1f}")
````

````python
# List comprehensions
# Instead of a loop to build a new list, you can write it in one line.

scores = [85, 92, 78, 95, 61, 88]

print("Original scores:", scores)

# Get only passing scores (>= 70)
passing = [s for s in scores if s >= 70]
print("Passing scores:", passing)

# Add a 5-point bonus to every score
boosted = [s + 5 for s in scores]
print("Boosted scores:", boosted)

# Add a 10-point bonus to failing scores (< 70)
failing_boosted = [s + 10 if s < 70 else s for s in scores]
print("Failing scores boosted:", failing_boosted)
````

## 6. Conditionals

Python uses `if`, `elif`, `else`. Note: **indentation defines the code block** (no `begin/end` or `{}`).


````python
score = 73

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score {score} → Grade {grade}")
````

````python
# Combining conditions
age = 17
has_id = True

if age >= 18 and has_id:
    print("Access granted")
elif age >= 18 or has_id:
    print("Partial access")
else:
    print("Access denied")
````

````python
# Checking membership with 'in'
valid_grades = ["A", "B", "C", "D", "F"]
user_input = "B"

if user_input in valid_grades:
    print(f"{user_input} is a valid grade")
else:
    print(f"{user_input} is not valid")
````

## 7. Loops

Python has `for` and `while` loops. `for` loops are used far more often in data work.


````python
# for loop over a list
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
````

````python
# range() generates numbers on the fly
for i in range(5):        # 0, 1, 2, 3, 4
    print(i, end=" ")
print()

for i in range(1, 11):    # 1 to 10
    print(i, end=" ")
print()

for i in range(0, 20, 5): # 0, 5, 10, 15 (step of 5)
    print(i, end=" ")
````

````python
# enumerate() gives you index AND value together
scores = [85, 92, 78, 95]

for i, score in enumerate(scores):
    print(f"Student {i}: {score}")
````

````python
# while loop
count = 0
while count < 5:
    print(f"Count: {count}")
    count += 1  # same as count = count + 1
````

## 8. Functions

Functions let you package reusable logic. In Python, you define them with `def`.


````python
def greet(name):
    """Returns a greeting message."""
    return f"Hello, {name}!"

print(greet("Alex"))
print(greet("World"))
````

````python
# Default parameter values
def calculate_tax(price, rate=0.15):
    """Calculate tax on a price. Default rate is 15% (VAT)."""
    return price * rate

print(calculate_tax(1000))        # uses default 15%
print(calculate_tax(1000, 0.10))  # override with 10%
````

````python
# Functions can return multiple values
def score_stats(scores):
    """Returns min, max and average of a list of scores."""
    return min(scores), max(scores), sum(scores) / len(scores)

low, high, avg = score_stats([85, 92, 78, 95, 61])
print(f"Low: {low}, High: {high}, Average: {avg:.1f}")
````

# 🏆 Mini Project: Student Grade Analyser

Time to put it all together! You're going to build a small grade analyser.

You have a list of student records (list of dictionaries). Your job is to write functions that analyse this data.

**Tasks:**
1. Print each student's name and their average score
2. Find and print the top student (highest average)
3. Count how many students are passing (average >= 50)
4. Bonus: Print a summary report at the end

Work through it step by step — don't try to do it all at once!

````python
# The data — don't change this cell
students = [
    {"name": "Alice",   "scores": [88, 92, 79, 95]},
    {"name": "Bob",     "scores": [45, 61, 53, 48]},
    {"name": "Carol",   "scores": [72, 68, 75, 80]},
    {"name": "David",   "scores": [95, 98, 91, 97]},
    {"name": "Eve",     "scores": [30, 42, 55, 38]},
    {"name": "Frank",   "scores": [65, 70, 68, 74]},
]
````

````python
# Task 1: Print each student's name and average score
# Expected output (roughly):
# Alice   — Average: 88.5
# Bob     — Average: 51.8
# ...

# Your code here
for student in students:
    ave = (sum(student['scores'])/len(student['scores']))
    student['average'] = ave
    print(f"{student['name']}\t-Average: {ave:.2f}")
````

````python
# Task 2: Find the top student (highest average score)
# Hint: keep track of the best name and best average as you loop

# Your code here
cur_best = [0, 0]
for student in students:
    if student['average'] > cur_best[0]:
        cur_best = [student['average'], students.index(student)]

print(f"{students[cur_best[1]]['name']} has the best score of {cur_best[0]:.2f}")
````

````python
# Cleaner modern way
from statistics import mean

best = max(students, key=lambda s: mean(s["scores"]))
avg = mean(best["scores"])

print(f"Top student: {best['name']} with average {avg:.1f}")
````

````python
# Task 3: Count how many students have an average >= 50

# Your code here
n = 0
for student in students:
    if student['average'] >= 50:
        n += 1
print(f'{n} students scored an average higher than 50')
````

````python
# Bonus: Write a function called print_report(students)
# that prints a nicely formatted summary like:
#
# ===== CLASS REPORT =====
# Alice   | Avg: 88.5 | Distinction
# Bob     | Avg: 51.8 | Pass
# Eve     | Avg: 41.3 | Fail
# ...
# Top student: David (95.3)
# Pass rate: 5/6 (83.3%)
# ========================

# Your code here

def print_report(students: list):
    cur_best = [0, 0]
    passed = 0
    print('===== CLASS REPORT =====')
    for student in students:
        if student['average'] >= 50:
            passed += 1
        if student['average'] > cur_best[0]:
            cur_best = [student['average'], student['name']]

        print(f'{student['name']} \t | Avg: {student['average']:.1f} \t | {classify_score(student["average"])}')

    print(f'Top student: {cur_best[1]} ({student['average']:.1f})')
    print(f'Pass rate: {passed}/{len(students)} ({passed/len(students):.1%}%)')
    print("========================")


print_report(students)
````

````python
best = max(students, key=lambda s: mean(s["scores"]))
avg = mean(best["scores"])
pass_rate = sum(1 for s in students if mean(s["scores"]) >= 50) / len(students) * 100

print('===== CLASS REPORT =====')


for s in students:
    avg_score = mean(s["scores"])
    status = "Distinction" if avg_score >= 85 else "Pass"
    print(f"{s['name']}    |   Avg:{avg_score:.1f}     | {status}")
print(f"Top student: {best['name']} ({avg:.1f})")
print(f"Pass rate: {pass_rate:.1f}%")
print('========================')
````
